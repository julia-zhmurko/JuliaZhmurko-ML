### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
# Код студента
a = np.array([1, 2])
print("Вектор а:", a)

b = np.array([2, 1])
print("Вектор b:", b)

print("\nСкалярний добуток векторів (a @ b):", a @ b)
print("Норма вектора a:", np.linalg.norm(a))
print("Норма вектора b:", np.linalg.norm(b))

cos_ang = (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))
print("\nКосинус кута між векторами а та b:", cos_ang)


print("\nКут у радіанах:", np.arccos(cos_ang))
print("Кут у градусах:", np.degrees(np.arccos(cos_ang)))
````

````python
# Код студента
arr = np.arange(1, 5, 1)
a = arr.reshape(2, 2)
print("Матриця A:")
print(a)

b = np.array([[2, 0], [1, 2]])
print("Матриця B:")
print(b)

#1
# Поелементний добуток (A * B) - Перемножуються елементи на однакових позиціях незалежно один від одного
# [1*2  2*0]   [2  0]
# [3*1  4*2] = [3  8]
print("\nПоелементний добуток матриць (A * B):")
print(a * b)

#2
# Матричне множення (A @ B) - Перемножуються РЯДКИ першої матриці на СТОВПЧИКИ другої
# Верхній-лівий:   (1*2 + 2*1) = 4
# Верхній-правий:  (1*0 + 2*2) = 4
# Нижній-лівий:    (3*2 + 4*1) = 10
# Нижній-правий:   (3*0 + 4*2) = 8
#   [ 4  4]
# = [10  8]
print("Матричний добуток матриць (A @ B):")
print(a @ b)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
arr = np.array([10, 20, 30, 20, 30, 40, 30, 40, 50])
a = arr.reshape(3, 3)
print("Матриця A:")
print(a)

mu = a.mean(axis = 0)
print("\nСереднє значення за стовпцями матриці А:", mu)

sigma = a.std(axis = 0)
print("Стандартне відхилення за стовпцями матриці А:", sigma)

z = (a - mu) / sigma
print("\nНормалізована матриця A_std:")
print(z)

print("\nПеревірка: Середнє значення за стовпцями нормалізованої матриці A_std:", z.mean(axis = 0))
print("Перевірка: Стандартне відхилення за стовпцями нормалізованої матриці A_std:", z.std(axis = 0))
````
