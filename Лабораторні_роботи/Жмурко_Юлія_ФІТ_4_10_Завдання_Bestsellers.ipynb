Відсортуйте за зростанням ціни всі книги, які потрапили в рейтинг у 2015 році і коштують дешевше за 8 доларів (використовуйте функцію `sort_values` ).

Яка книга остання у відсортованому списку?

```python
filtered_books = df[(df['year'] == 2015) & (df['price'] < 8)]

sorted_books = filtered_books.sort_values(by = 'price', ascending = True)

print("Відсортований список книг 2015 року дешевших за 8 доларів:")
display(sorted_books)

last_book = sorted_books.iloc[-1]

print("\nОстання книга у відсортованому списку:")
print(f"Назва: {last_book['name']}")
print(f"Автор: {last_book['author']}")
print(f"Ціна: ${last_book['price']}")
```

**У відсортованому списку останньою є книга Old School (Diary of a Wimpy Kid #10).**

Створити новий датафрейм, який вміщатиме кількість книг для кожного з авторів (використовуйте функції `groupby` і `agg` , для підрахунку кількості використовуйте `count` ). Не беріть усі стовпці, виберіть тільки потрібні.

* Якої розмірності вийшла таблиця?
* У якого автора найбільше книг?
* Скільки книг у цього автора?

```python
author_books_count = df.groupby('author')['name'].agg(['count']).reset_index()

author_books_count.columns = ['author', 'books_count']

print("Датафрейм кількості книг для кожного автора:")
display(author_books_count)

shape = author_books_count.shape
print(f"\nРозмірність таблиці: {shape} (рядків: {shape[0]}, стовпців: {shape[1]})")

top_author_row = author_books_count.sort_values(by='books_count', ascending=False).iloc[0]

print(f"Автор з найбільшою кількістю книг: {top_author_row['author']}")
print(f"Кількість книг у цього автора: {top_author_row['books_count']}")
```

* **Таблиця вийшла розмірності (248, 2).**

* **Найбільше книг у автора Jeff Kinney.**

* **У автора Jeff Kinney всього 12 книг.**
